```python
from db_manager_v2 import allocate_from_security_to_investment
from sqlalchemy import select
from db_manager_v2 import Security
import time

sol_id = 0
amount_to_gas = 53286757          # ← Change this if you want more or less SOL as Gas

sol_savings = session.execute(
    select(Security).where(Security.id == sol_id)
).scalar_one()

if sol_savings.amount >= amount_to_gas:
    sol_savings.amount -= amount_to_gas

    gas_sec = Security(
        parent_id=sol_savings.parent_id,
        amount=amount_to_gas,
        purchase_price_usdc=0.0,
        purchase_time_ms=int(time.time() * 1000),
        buy_fee_native_lamports=0,
        buy_fee_usdc=0.0,
        buy_tx_id=f"to_gas_{sol_id}",
        isClosed=False,
        isSavings=False,
        isGas=True,
    )
    session.add(gas_sec)
    session.commit()

    print(f"✅ Moved {amount_to_gas} lamports SOL to Gas (new Security id={gas_sec.id})")
else:
    print("Not enough SOL in Savings")
```
